```python
#Vectors are objects that have both a magnitude (absolute size) and a direction
#Example an arrow has a line representing length (magnitude) and an arrow that
#represents direction
#The direction of the vector is from its tail to its head
#A one dimensonal matrix is called a vector
#Vectors are categorized into Row vectors, and column vectors
#Create a vector using NumPy
#Import NumPy Library
import numpy as np
#Create a 1-D list (horizontal)
list1 = [1, 2, 3]
#Create a 1-D list (vertical)
list2 = [[10],
         [20],
         [30]]
#Create a vector1
#Vector as row
vector1 = np.array(list1)
#Create vector2
#Vector as column
vector2 = np.array(list2)

#Syntax np.array(list)
#Alternate ways to create a vector
# np.arange(10)
# np.zeros(4, 1)
```

```python
#Mathematical Implementation of Vectors

#Import numpy
import numpy as np

#Create 1-D list (Horizontal)
list1 = [11, 61, 19]
#Create 1-D list (Horizontal)
list2 = [31, 34, 53]
#Create first vector
vector1 = np.array(list1)
#Print first Vector
print(vector1)
#Create second vector
vector2 = np.array(list2)
#Print vector 2
print(vector2)
```

```python
from __future__ import division
#add both vectors
# A+B= (a1+b1, a2+b2, a3+b3)
addition = vector1 + vector2
#Print addition vector
print("Vector Addition :" + str(addition))

#Subtract the vectors
#a-b = (a1-b1, a2-b2, a3-b3)
subtraction = vector1 - vector2
#Print subtraction vector
print("Vector Subtraction :" + str(subtraction))

#Multiply the vectors
#a*b = (a1*b1, a2*b2, a3*b3)
multiplication = vector1 * vector2
#Print multiplication vector
print("Vector Multiplication :" + str(multiplication))

#Dividing both vectors
#a/b = (a1/b1, a2/b2, a3/b3)
division = vector1 / vector2
#Print division vector
print("Vector Division :" + str(division))
```

```python
#Dot products of two 1-D vectors
#also known as inner product
#Vectors must be the same size
#The result of a dot product is a SCALAR
#multiplies element by element

#Dot product using np.dot()
#getting dot product of both vectors
#a
```

```python
#Dot product using np.dot()
#getting dot product of both vectors
dot_product = np.dot(vector1, vector2)
print("Dot Product: " + str(dot_product))

#Dot product using the np.inner() function
dot_product = np.inner(vector1, vector2)
print("The Dot Product of x and y is: " + str(dot_product))

#Dot product using an explicit @ operator
print("The dot product of vector1 and vector 2 is: ", vector1 @ vector2)
```

```python
#Scalar Product of two 1-D vectors
#print vector1
print("Vector 1: " + str(vector1))

#scalar value
scalar = 2
#print scalar value
print("Scalar Value: " + str(scalar))

#Get scalar multiplication value
#a * v = (a*v1, a*v2, a*v3)
scalar_nul = vector1 * scalar

#print dot product
print("Scalar Multiplication: " + str(scalar_nul))
```
