```python
#Eigen vector does not change direction (span) during transformation
#refer to intro to ml week2 slide 123 for the mathematical equation
#representing an Eigenvector/Eigenvalue for a square matrix
#Eigen Values- the SCALAR values that associate to the vector

#creating a matrix for reference
import numpy as np
my_matrix = np.array([[-6, 3],
                      [4, 5]])
```

```python
import numpy as np

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(my_matrix)

print("Eigenvalues:")
print(eigenvalues)
print("\nEigenvectors:")
print(eigenvectors)
```
