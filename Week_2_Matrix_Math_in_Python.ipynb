```python
# Creating a Matrix

# Loading library: importing numpy package
import numpy as np

# Create a matrix
matrix = np.array([[1, 2],
                   [1, 2],
                   [1, 2]])
matrix
```

```python
#Add two matrix
#creating matrix 1
matrix_a = np.array([[1, 1, 1],
                     [1, 1, 1],
                     [1, 1, 2]])
# matrix 2
matrix_b = np.array([[1, 3, 1],
                     [1, 3, 1],
                     [1, 3, 8]])

#add the 2 matrixes
np.add (matrix_a, matrix_b)

```

```python
#subtract the 2 matrixes with the following command
matrix_a - matrix_b
```

```python
#create matrix a
matrix_a= np.array([[1, 1, 1],
                    [1, 1, 1],
                    [1, 1, 2]])

#create matrix b
matrix_b= np.array([[1, 3, 1],
                    [1, 3, 1],
                    [1, 3, 8]])
#subtrat them with this function
np.subtract(matrix_a, matrix_b)
```

```python
#Matrix multiplication

#multiply by using np.dot

np.dot(matrix_a, matrix_b)
```

```python
#Matrix multiplication

#multiply by using @ operator

matrix_a @ matrix_b
```

```python
#Matrix multiplication

#multiply by using element wise: Element-wise (Hadamard) Multiplication:
# You multiply each entry of one matrix by the corresponding entry in the same
#position in the other matrix.

matrix_a * matrix_b
```

```python
#Flip a matrix over its diagonal using the TRANSPOSE operation...
# uses the first created one as reference
matrix.T
```

```python
#Change the signs of values across the main diagonal of a matrix
# by using the INVERSE function
np.linalg.pinv(matrix)
```
